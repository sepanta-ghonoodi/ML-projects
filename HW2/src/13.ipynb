```python
import numpy as np


def F_system(w):

    x, y, lam = w[0], w[1], w[2]

    exp_term = np.exp(x**2 + y**2)

    f1 = 2 * x * exp_term + lam
    f2 = 2 * y * exp_term + lam
    f3 = x + y - 1

    return np.array([f1, f2, f3])
```

```python
def jacobian(w):
    x, y, lam = w[0], w[1], w[2]

    exp_term = np.exp(x**2 + y**2)

    df1_dx = 2 * exp_term + 2 * x * (2 * x) * exp_term
    df1_dy = 2 * x * (2 * y) * exp_term
    df1_dl = 1

    df2_dx = 2 * y * (2 * x) * exp_term
    df2_dy = 2 * exp_term + 2 * y * (2 * y) * exp_term
    df2_dl = 1

    df3_dx = 1
    df3_dy = 1
    df3_dl = 0

    J = np.array([
        [df1_dx, df1_dy, df1_dl],
        [df2_dx, df2_dy, df2_dl],
        [df3_dx, df3_dy, df3_dl]
    ])

    return J
```

```python
analytic_x = 0.5
analytic_y = 0.5
analytic_lam = -np.exp(0.5)
w_star = np.array([analytic_x, analytic_y, analytic_lam])

print(f"Target Analytic Solution w*: {w_star}")
print("-" * 50)

w = np.array([0.0, 0.0, 0.0])
iterations = 10
tol = 1e-8

print(f"{'Iter':<5} | {'x':<10} | {'y':<10} | {'lambda':<10} | {'Error ||w - w*||'}")
print("-" * 65)

for k in range(iterations):
    F_val = F_system(w)
    J_val = jacobian(w)

    if np.linalg.norm(F_val) < tol:
        print(f"\nConverged at iteration {k}")
        break
    delta = np.linalg.solve(J_val, F_val)
    w = w - delta

    error = np.linalg.norm(w - w_star)

    print(f"{k+1:<5} | {w[0]:.8f} | {w[1]:.8f} | {w[2]:.8f} | {error:.2e}")

print("-" * 65)
print(f"Final Solution w: {w}")
```
