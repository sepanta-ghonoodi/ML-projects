```python
import numpy as np
import matplotlib.pyplot as plt

n_samples = 100
p1 = 0.6
p2 = 0.4

samples1 = np.random.multivariate_normal(
    np.array([4, 11]), 3 * np.identity(2), n_samples)
samples2 = np.random.multivariate_normal(
    np.array([10, 3]), 3 * np.identity(2), n_samples)

all_samples = np.vstack((samples1, samples2))
x_min, x_max = -5, 20
y_min, y_max = -5, 20

x_boundary = np.linspace(x_min, x_max, 100)

y_boundary = 12/16 * x_boundary + 28/16 + 6/16 * np.log(p1 / p2)

plt.figure()

plt.scatter(samples1[:, 0], samples1[:, 1], c='blue',
            alpha=0.7, label='Class 1 Samples (ω1)')
plt.scatter(samples2[:, 0], samples2[:, 1], c='red',
            alpha=0.7, label='Class 2 Samples (ω2)')

plt.scatter(4, 11, c='darkblue', marker='x',
            label='Mean 1 (4, 11)')
plt.scatter(10, 3, c='darkred', marker='x', label='Mean 2 (10, 3)')

plt.plot(x_boundary, y_boundary, c='green', linestyle='--',
         linewidth=2, label='Decision Boundary (g1=g2)')

plt.title('Sampled Points and Bayesian Decision Boundary')
plt.xlabel('x1 (Feature 1)')
plt.ylabel('x2 (Feature 2)')

plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)


plt.gca().set_aspect('equal', adjustable='box')

plt.savefig('gaussian_decision_boundary.png')
```
